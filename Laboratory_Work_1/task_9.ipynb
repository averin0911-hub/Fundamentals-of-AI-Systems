```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# В переменную uniform_and_not загружена таблица из файла uniform_and_not_data
uniform_and_not = pd.read_csv('uniform_and_not_data.csv')
fig, axes = plt.subplots(2, 3, figsize=(15, 8))
for i, ax in zip("ABCDE", axes.flat):
    sns.histplot(data=uniform_and_not, x=i, ax=ax, bins=10)
    mean_y = len(uniform_and_not) / 10
    ax.axhline(mean_y, color="red", linestyle="--")
plt.show()
```
