```python
import pandas as pd
```

```python
movies = pd.read_csv("movies_data.csv")
movies = movies[~movies["Год производства"].isna()]
```

```python
len(movies[(movies["Страна производства"] == "Франция") & (movies["Жанр фильма"] == "Драма")])
```

```python
len(movies[(movies["Страна производства"] == "Россия") & (movies["Год производства"] > 2010)])
```

```python
len(movies[(movies["Страна производства"] == "Россия") | \
            (movies["Год производства"] < 2000) & (movies["Жанр фильма"] == "Триллер")])
```
